```python
#@title 🚀 Khởi động Rich Future Voice { display-mode: "form" }
REPO_URL = "https://github.com/Cosibility/rich-future-voice.git"

import os, pathlib, shutil, subprocess, sys

repo_dir = pathlib.Path("/content/RichFutureVoice")
if repo_dir.exists():
    shutil.rmtree(repo_dir)
subprocess.run(["git", "clone", "--depth", "1", REPO_URL, str(repo_dir)], check=True)
source_revision = subprocess.check_output(["git", "-C", str(repo_dir), "rev-parse", "HEAD"], text=True).strip()
print(f"Mã nguồn Rich Future: {source_revision}")

env = os.environ.copy()
env["RICH_FUTURE_SOURCE_URL"] = REPO_URL.removesuffix(".git")
_setup = subprocess.run([sys.executable, "scripts/colab_one_click.py"], cwd=repo_dir, env=env, check=True)

from html import escape
from google.colab import output
from IPython.display import HTML, display

proxy_url = output.eval_js("google.colab.kernel.proxyPort(3900)")
safe_url = escape(str(proxy_url), quote=True)
display(HTML(
    '<div style="margin:16px 0;padding:18px;border:1px solid #14b8a6;border-radius:12px;background:#071827">'
    '<div style="margin-bottom:12px;color:#f0fdfa;font-weight:700">Phiên riêng tư của bạn</div>'
    '<div style="margin-bottom:14px;color:#b8d8df">Không chia sẻ URL ứng dụng. Mỗi người dùng phải tự chạy notebook trong tài khoản Colab riêng.</div>'
    f'<a href="{safe_url}" target="_blank" rel="noopener noreferrer" '
    'style="display:inline-block;padding:12px 20px;border-radius:8px;background:#14b8a6;'
    'color:#03111d;font-weight:700;text-decoration:none">MỞ RICH FUTURE VOICE ↗</a>'
    f'<div style="margin-top:10px;color:#b8d8df">{safe_url}</div></div>'
))
print("✓ Bấm nút phía trên hoặc dùng ứng dụng được nhúng bên dưới.")
output.serve_kernel_port_as_iframe(3900, width="100%", height="900")
```
