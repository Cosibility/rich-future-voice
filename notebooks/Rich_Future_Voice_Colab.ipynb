```python
#@title 🚀 Khởi động Rich Future Voice { display-mode: "form" }
REPO_URL = "https://github.com/Cosibility/rich-future-voice.git"

import os, pathlib, subprocess, sys

repo_dir = pathlib.Path("/content/RichFutureVoice")
if (repo_dir / ".git").is_dir():
    subprocess.run(["git", "-C", str(repo_dir), "pull", "--ff-only"], check=True)
else:
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, str(repo_dir)], check=True)

env = os.environ.copy()
env["RICH_FUTURE_SOURCE_URL"] = REPO_URL.removesuffix(".git")
subprocess.run([sys.executable, "scripts/colab_one_click.py"], cwd=repo_dir, env=env, check=True)
```
